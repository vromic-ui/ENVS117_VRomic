## Part A. Read and clean the data (20 pts)

**A1.** Read the file inside `try` / `except FileNotFoundError`. Store all lines in a list.

```python
# Your code here
try: 
    with open("south_bay_pm25.csv") as f:
        lines = [line.rstrip() for line in f]
except FileNotFoundError: 
    print("No such file exist")
```

**A2 and A3.** Write a function `parse_line(line)` that returns a dictionary for a valid row and `None` for an unusable one.

A row is unusable if it does not have 6 fields, if a number cannot be converted to `float`, or if PM2.5 is negative (the logger uses `-999` as an error code). Standardize station IDs (remove extra spaces, uppercase). The dictionary keys should be `station_id`, `name`, `lat`, `lon`, `timestamp`, `pm25`.

```python
def parse_line(line):
    """Parse one data line. Return a dict, or None if the line is unusable."""
    fields = line.strip().split(',')
    if len(fields) != 6:
        return None 
    raw_id, name, raw_lat, raw_lon, timestamp, raw_pm25 = fields
    station_id = raw_id.strip().upper()

    try:
        lat = float(raw_lat.strip())
        lon = float(raw_lon.strip())
        pm25 = float(raw_pm25.strip())
    except ValueError:
        return None

    if pm25<0:
        return None
        
    return {
        'station_id': station_id, 
        'name': name.strip(), 
        'lat': lat, 
        'lon': lon, 
        'timestamp': timestamp.strip(),
        'pm25': pm25
    }
    pass
```

**A4.** Loop over the data lines (skip the header), skip blank lines, and build a list `records` of valid dictionaries. Print each skipped line, then print how many data lines were read, kept and skipped.

```python
# Your code here
#create empty list and empty counts
records = []
read = 0
kept = 0
skipped = 0
#start of for loop 
for line in lines[1:]:
   
    if not line.strip():
        continue
    read += 1
    record = parse_line(line)

    if record is not None:
        records.append(record)
        kept += 1
    else: 
        print(f"Skipped Line: {line}")
        skipped += 1 

print(f"Total read: {read}")
print(f"Total kept: {kept}")
print(f"Total skipped: {skipped}")    
```

## Part C. Functions and station summaries (25 pts)

**C1.** `pm25_category(pm)`: EPA AQI category for a 24-hour PM2.5 value (ug/m3). Breakpoints: Good 0 to 9.0, Moderate to 35.4, Unhealthy for Sensitive Groups to 55.4, Unhealthy to 125.4, Very Unhealthy to 225.4, Hazardous above.

```python
# Your code here
def pm25_category(pm):
    #sorts pm25 values into air quality scores
    if pm<=9.0:
        return "Good"
    if 9.0<=pm<=35.4:
        return "Moderate"
    if 35.4<=pm<=55.4:
        return "Unhealthy for Sensitive Groups"
    if 55.4<=pm<=125.4:
        return "Unhealthy"
    if 125.4<=pm<=225.4:
        return "Very Unhealthy"
    if pm>=225.4:
        return "Hazardous"
pm25_category(300)
     
```

**C2.** `haversine(lat1, lon1, lat2, lon2, radius=6371.0)`: great-circle distance (km by default).

```python
# Your code here
def haversine(lat1, lon1, lat2, lon2, radius=6371.0):
    #calculates great circle distance in km 
    dlat = radians(lat2 - lat1)
    dlon = radians(lon2 - lon1)
    a = sin(dlat / 2) ** 2 + cos(radians(lat1)) * cos(radians(lat2)) * sin(dlon / 2) ** 2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))
    return radius * c
```

**C3.** `summarize(readings, threshold=35.4)`: return the mean, the peak and the number of readings above the threshold.

```python
# Your code here
def summarize(readings, threshold=35.4):
    #returns the mean, peak, and the number of readings above the threshold 
    mean = sum(readings)/len(readings)
    peak = max(readings)
   
    readings_above_threshold = 0
    for reading in readings:
        if reading>=threshold:
            readings_above_threshold += 1
            
    return mean, peak, readings_above_threshold
```

**C4 and C5.** Group the readings by station (a dictionary of lists). Then, for each station, compute: number of valid hours, 24-hour mean, peak, hours above 35.4 ug/m3, AQI category of the mean, and distance to the fire (km). Store one dictionary per station in a list called `summary`, and print a neatly aligned table using f-string widths (e.g. `{sid:<7}`, `{mean:6.1f}`).

```python
# Your code here

#Group readings by station ID into a dictionary of lists
station_readings = {}
for record in records:
    station_id = record['station_id']
    if station_id not in station_readings:
        station_readings[station_id] = []
    station_readings[station_id].append(record['pm25'])

#Build summary list
summary = []

for station_id, readings in station_readings.items():
    mean, peak, hours_above = summarize(readings, threshold=35.4)
    valid_hours = len(readings)
    pm25_cat = pm25_category(mean) 
    
    lat, lon = stations[station_id]['location']
    dist_km = haversine(lat, lon, FIRE_LAT, FIRE_LON) 
    
    station_summary = {
        "station_id": station_id,
        "valid_hours": valid_hours,
        "mean": mean,
        "peak": peak,
        "hours_above_35.4": hours_above,
        "pm_category": pm25_cat,
        "distance_km": dist_km
    }
    summary.append(station_summary)

#Print formatted summary table
print(f"{'ID':<10} {'Hours':<7} {'Mean':>7} {'Peak':>7} {'>35.4':>7} {'Dist(km)':>10}  {'AQI Category'}")
print("-" * 75)

for s in summary:
    print(
        f"{s['station_id']:<10} "
        f"{s['valid_hours']:<7d} "
        f"{s['mean']:>7.1f} "
        f"{s['peak']:>7.1f} "
        f"{s['hours_above_35.4']:>7d} "
        f"{s['distance_km']:>10.1f}  "
        f"{s['pm_category']}"
    )
```

**D3.** Use a `while` loop to find the first hour when PM2.5 at `SCU-01` exceeded 35.4 ug/m3.

```python
# Your code here
scu_records = [r for r in records if r['station_id'] == 'SCU-01']

i = 0
first_exceedance_time = None

while i<len(scu_records):
    if scu_records[i]['pm25']> 35.4:
        first_exceedance_time = scu_records[i]['timestamp']
        break
    i += 1

print(f"First exceedance time at SCU-01: {first_exceedance_time}")
        
```

**D4.** Compare the average of station means within 40 km of the fire with those farther away.

```python
# Your code here
close_means = []
far_means = []

for s in summary:
    if s['distance_km']<=40:
        close_means.append(s['mean'])
    else:
        far_means.append(s['mean'])

avg_close = sum(close_means)/len(close_means)
avg_far = sum(far_means)/len(far_means)

print(f"The average of the station means within 40 km is {avg_close:.1f} whereas the average of the station means that are farther away is {avg_far:.1f}")
              
```
